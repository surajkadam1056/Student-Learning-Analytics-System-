```python
attendance = pd.read_csv("attendance.csv")

sql = """
INSERT INTO attendance
(attendance_id, student_id, course_id, attendance_date, status)
VALUES (%s,%s,%s,%s,%s)
"""

for row in attendance.itertuples(index=False):
    cursor.execute(sql, tuple(row))

connection.commit()

print("Attendance Uploaded")
```

```python
import pandas as pd

assignment_submissions = pd.read_csv("assignment_submissions.csv")

# Convert NaN to None
assignment_submissions = assignment_submissions.where(
    pd.notnull(assignment_submissions),
    None
)

sql = """
INSERT INTO assignment_submissions
(submission_id, assignment_id, student_id,
submission_date, marks_obtained, submission_status)
VALUES (%s, %s, %s, %s, %s, %s)
"""

data = list(assignment_submissions.itertuples(index=False, name=None))

cursor.executemany(sql, data)

connection.commit()

print("Uploaded Successfully")
```

```python
cursor.close()
connection.close()
```

```python
import pandas as pd

exam_results = pd.read_csv("exam_results.csv")

sql = """
INSERT INTO exam_results
(result_id, exam_id, student_id,
marks_obtained, grade, pass_fail)
VALUES (%s, %s, %s, %s, %s, %s)
"""

# DataFrame ko tuple list me convert karo
data = list(exam_results.itertuples(index=False, name=None))

# Ek hi baar me multiple rows insert
cursor.executemany(sql, data)

connection.commit()

print(f"{cursor.rowcount} Exam Results Uploaded Successfully")

cursor.close()
connection.close()
```
